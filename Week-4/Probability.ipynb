```python
# 2. Conditional probability

# bag: 5 red, 3 blue. P(second red | first red)
p = 4/7
print(p)
```

```python
# 3. Bayes' theorem

p_disease = 0.01
p_pos_given_disease = 0.99
p_pos_given_no_disease = 0.05

p_positive = p_pos_given_disease*p_disease + p_pos_given_no_disease*(1-p_disease)
p_disease_given_pos = (p_pos_given_disease * p_disease) / p_positive
print(p_disease_given_pos)   
```

```python
# 4. Normal distribution

from math import erf, exp, pi, sqrt

def norm_pdf(x, loc=0, scale=1):
    z = (x - loc) / scale
    return (1 / (scale * sqrt(2 * pi))) * exp(-0.5 * z * z)

def norm_cdf(x, loc=0, scale=1):
    z = (x - loc) / (scale * sqrt(2))
    return 0.5 * (1 + erf(z))

print(norm_pdf(0, loc=0, scale=1))   # density at x=0
print(norm_cdf(1) - norm_cdf(-1))    # P(-1 < x < 1) ~ 0.68
```

```python
# 5. Binomial distribution

from math import comb

n = 10
k = 6
p = 0.5

probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
print(probability)   # P(exactly 6 heads in 10 flips)
```
