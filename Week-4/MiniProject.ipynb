```python
# Mini Project: Setup + Cleaning

# 1. Load data

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine

data = load_wine(as_frame=True)

df = data.frame
df.head()
```

```python
# 2. Check shape and structure

print("Shape:", df.shape)
df.info()
```

```python
# 3. Check data types

df.dtypes
```

```python
# 4. Check missing values

df.isnull().sum()
```

```python
# 5. Check duplicates

dupes = df.duplicated().sum()
print("Duplicate rows:", dupes)
```

```python
# Drop duplicates
df = df.drop_duplicates()
print("Shape after dropping duplicates:", df.shape)
```

```python
# 6. Check basic statistics (sanity check for outliers/errors)

df.describe()
```

```python
# 7. Check target class balance

df['target'].value_counts()
```

```python
# 8. Rename target for clarity

df = df.rename(columns={'target': 'cultivar'})
df.head()
```

```python
# EDA 

# 9.Univariate analysis — distribution of alcohol content

plt.figure(figsize=(6,4))
plt.hist(df['alcohol'], bins=20, color='steelblue', edgecolor='black')
plt.title("Distribution of Alcohol Content")
plt.xlabel("alcohol"); plt.ylabel("count")
plt.show()

# ---> Insight: alcohol content is roughly spread between 11 and 15, with a mild concentration around 12-13.
```

```python
# 10. Univariate analysis — boxplot of color intensity

plt.figure(figsize=(4,5))
plt.boxplot(df['color_intensity'])
plt.title("Color Intensity Spread")
plt.ylabel("color_intensity")
plt.show()
# ---> Insight: a few high outliers exist above ~10, but the bulk of samples sit between 2 and 6.
```

```python
# Ensure dataframe is properly prepared
if 'cultivar' not in df.columns:
    df = df.rename(columns={'target': 'cultivar'})

# 11. Bivariate analysis — alcohol vs color intensity by cultivar

plt.figure(figsize=(6,5))
for cultivar in df['cultivar'].unique():
    subset = df[df['cultivar'] == cultivar]
    plt.scatter(subset['alcohol'], subset['color_intensity'], label=f'cultivar {cultivar}', alpha=0.7)
plt.xlabel("alcohol"); plt.ylabel("color_intensity")
plt.title("Alcohol vs Color Intensity by Cultivar")
plt.legend()
plt.show()
```

```python
# 12.  Correlation heatmap

plt.figure(figsize=(10,8))
corr = df.drop(columns='cultivar').corr()
sns.heatmap(corr, cmap='coolwarm', center=0, annot=False)
plt.title("Feature Correlation Heatmap")
plt.show()
```
