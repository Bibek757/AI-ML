```python
# 1. Matrix multiplication (2x2)

import numpy as np
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
print(A @ B)
```

```python
# 2. Matrix multiplication (3x3)

A3 = np.array([[1,0,2],
               [0,1,1],
               [2,1,0]])
B3 = np.array([[1,2,0],
               [0,1,1],
               [1,0,1]])
print(A3 @ B3)
```

```python
# 3. Determinant (2x2)
# Measures how much a matrix scales area. det=0 means it collapses space (no inverse).

A = np.array([[1, 2],
              [3, 4]])
print(np.linalg.det(A)) 
 
```

```python
# 4. Determinant (3x3)

A3 = np.array([[1,0,2],
               [0,1,1],
               [2,1,0]])
print(np.linalg.det(A3))
```

```python
# 5. Inverse
# A @ A_inv = Identity matrix. Only exists if det != 0.

A = np.array([[4, 7],
              [2, 6]])
A_inv = np.linalg.inv(A)
print(A_inv)
print(A @ A_inv)   # should be identity
```

```python
# 6. Singular matrix (no inverse)

# det=0 means matrix has no inverse (throws error).

S = np.array([[1, 2],
              [2, 4]])
print(np.linalg.det(S))   # 0.0
try:
    np.linalg.inv(S)
except np.linalg.LinAlgError as e:
    print("Error:", e)
```
