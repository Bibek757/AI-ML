```python
## 1. Load Titanic dataset, print shape and column names
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")
print(df.shape)
print(df.columns.tolist())
```

```python
# 2. Print dtypes, identify which columns are numeric vs object
print(df.dtypes)

numeric_cols = df.select_dtypes(include="number").columns.tolist()
object_cols = df.select_dtypes(include="object").columns.tolist()

print("Numeric:", numeric_cols)
print("Object:", object_cols)
```

```python
# 3. Run .info() and note which columns have missing values
df.info()
```

```python
# 4. Run .describe() — find a column with a surprising min or max
print(df.describe())
```

```python
# 5. Count total missing values per column, sort descending
missing = df.isnull().sum().sort_values(ascending=False)
print(missing)
```

```python
# 6. Check for duplicate rows, print them if any exist
dupe_count = df.duplicated().sum()
print("Duplicate rows:", dupe_count)

if dupe_count > 0:
    print(df[df.duplicated()])
```

```python
# 7. Pick one categorical column, run .unique() and .value_counts()
print(df["embarked"].unique())
print(df["embarked"].value_counts())
```

```python
# 8. Filter and print all rows where 'age' is missing
missing_age = df[df["age"].isnull()]
print(missing_age.shape)
print(missing_age.head())
```
