```python
# Step 1: Load libraries and dataset

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('titanic_cleaned.csv')
numeric_cols = ['age', 'fare', 'sibsp', 'parch', 'pclass']
df.head()
```

```python
# Step 2: Set plot style
# Consistent style across all charts — do this once, reuse everywhere.

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

```python
# Step 3: Histogram — Age distribution

sns.histplot(df['age'], bins=30, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# Step 4: Histogram — Fare distribution

sns.histplot(df['fare'], bins=30, kde=True)
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.show()
```

```python
# Step 5: Histogram — sibsp and parch (family size features)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df['sibsp'], bins=range(0, 9), ax=axes[0])
axes[0].set_title('Siblings/Spouses Aboard')
sns.histplot(df['parch'], bins=range(0, 7), ax=axes[1])
axes[1].set_title('Parents/Children Aboard')
plt.tight_layout()
plt.show()
```

```python
# Step 6: Box plot — Age

sns.boxplot(x=df['age'])
plt.title('Age Box Plot')
plt.show()
```

```python
# Step 7: Box plot — Fare

sns.boxplot(x=df['fare'])
plt.title('Fare Box Plot')
plt.show()
```

```python
# Step 8: Box plot — Fare by class
# Grouped box plot to check if fare outliers are explained by passenger class.

sns.boxplot(x='class', y='fare', data=df, order=['First', 'Second', 'Third'])
plt.title('Fare by Class')
plt.show()
```

```python
# Step 9: Box plot — Age by survival
# Check if age distribution differs between survivors and non-survivors.

sns.boxplot(x='survived', y='age', data=df)
plt.title('Age by Survival (0 = Died, 1 = Survived)')
plt.show()
```
