```python
# Step 1: Load the dataset

import pandas as pd

df = pd.read_csv('titanic_cleaned.csv')
df.head()
```

```python
# Step 2: Inspect structure

print(df.shape)      # rows, columns
print(df.dtypes)     # data type per column
```

```python
# Step 3: Check for missing values

df.isnull().sum()
```

```python
# Step 4: Separate numeric vs categorical columns

numeric_cols = ['age', 'fare', 'sibsp', 'parch', 'pclass']
categorical_cols = ['sex', 'class', 'embarked', 'who']
```

```python
# Step 5: Compute descriptive stats on numeric columns

stats = df[numeric_cols].describe().T
stats
```

```python
# Step 6: Add variance and mode manually

stats['variance'] = df[numeric_cols].var()
stats['mode'] = df[numeric_cols].mode().iloc[0]
stats
```

```python
# Step 7: Categorical breakdown

for col in categorical_cols:
    print(df[col].value_counts())
    print()
```

```python
# Step 8: Cross-tab survival rate by key groups

print("Survival rate by sex:")
print(df.groupby('sex')['survived'].mean())
print()
print("Survival rate by class:")
print(df.groupby('class')['survived'].mean())
```

```python
# Step 9: Save output
stats.to_csv('stats_summary.csv')
print("Saved stats_summary.csv")
```
